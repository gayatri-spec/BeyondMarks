```python
# BeyondMarks - Skill Gap Priority

career_skills = [
    "Excel",
    "Power BI",
    "Tableau",
    "SQL",
    "Python",
    "SAS",
    "Alteryx"
]

# Example student skills
student_skills = {
    "Excel": 1,
    "Power BI": 1,
    "Tableau": 0,
    "SQL": 1,
    "Python": 1,
    "SAS": 0,
    "Alteryx": 0
}

# O*NET status from our extracted data
skill_status = {
    "Excel": ("Y", "Y"),
    "Power BI": ("Y", "Y"),
    "Tableau": ("Y", "Y"),
    "SQL": ("Y", "Y"),
    "Python": ("Y", "Y"),
    "SAS": ("Y", "Y"),
    "Alteryx": ("Y", "N")
}

print("\n===== BEYONDMARKS SKILL GAP =====")

for skill in career_skills:

    if student_skills[skill] == 0:

        hot = skill_status[skill][0]
        demand = skill_status[skill][1]

        if hot == "Y" and demand == "Y":
            priority = "HIGH"

        elif hot == "Y" or demand == "Y":
            priority = "MEDIUM"

        else:
            priority = "LOW"

        print(
            f"{skill:<15} → {priority}"
        )
```
