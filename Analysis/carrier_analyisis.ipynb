```python
import pandas as pd

# Load O*NET software skills
software_df = pd.read_csv(
    "../data/raw/software_skills.csv"
)

# Career paths
careers = {
    "Business Intelligence Analyst": "15-2051.01",
    "Data Scientist": "15-2051.00",
    "Operations Research Analyst": "15-2031.00",
    "Computer Systems Analyst": "15-1211.00"
}

# Technologies we want to compare
technology_patterns = [
    "Microsoft Excel",
    "Microsoft Power BI",
    "Tableau",
    "Python",
    "R",
    "SAS",
    "Alteryx",
    "MATLAB",
    "SPSS",
    "Hadoop",
    "Spark",
    "AWS",
    "Azure"
]

career_skill_data = []

for career, code in careers.items():

    career_data = software_df[
        software_df["O*NET-SOC Code"] == code
    ]

    result = career_data[
        career_data["Workplace Example"].isin(
            technology_patterns
        )
    ]

    # Remove duplicate technologies
    result = result.drop_duplicates(
        subset=["Workplace Example"]
    )

    for _, row in result.iterrows():

        career_skill_data.append({
            "Career": career,
            "Technology": row["Workplace Example"],
            "Hot Technology": row["Hot Technology"],
            "In Demand": row["In Demand"]
        })


# Create final table
career_skills_df = pd.DataFrame(
    career_skill_data
)

print("\n===== CLEAN CAREER SKILL TABLE =====\n")

print(
    f"{'Career':<35}"
    f"{'Technology':<25}"
    f"{'Hot':<8}"
    f"{'In Demand':<10}"
)

print("-" * 78)

for _, row in career_skills_df.iterrows():

    print(
        f"{str(row['Career']):<35}"
        f"{str(row['Technology']):<25}"
        f"{str(row['Hot Technology']):<8}"
        f"{str(row['In Demand']):<10}"
    )

print("\nTotal records:", len(career_skills_df))
```
